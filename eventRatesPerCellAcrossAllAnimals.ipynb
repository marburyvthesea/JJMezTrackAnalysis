```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
```

```python
def natural_mouse_key(mouse_name):
    m = re.fullmatch(r"m(\d+)", str(mouse_name))
    if m:
        return int(m.group(1))
    return str(mouse_name)
```

```python
base_dir = Path("/scratch/jma819/CaliAli_linearTrackData")
window_len = 20

# load per-mouse saved data
mouse_eventrate_files = sorted(
    base_dir.glob(f"m*_analysis/eventsPerSecond_with_velocity_window{window_len}.csv"),
    key=lambda p: natural_mouse_key(p.parent.name.replace("_analysis", ""))
)

events_per_second_with_velocity_by_mouse = {}

for csv_path in mouse_eventrate_files:
    print(csv_path)
    print('reading event rate file')
    mouse = csv_path.parent.name.replace("_analysis", "")
    df = pd.read_csv(csv_path, index_col=0)
    df.index.name = "window_start_idx"
    events_per_second_with_velocity_by_mouse[mouse] = df
    print(mouse, df.shape)

# compute mean event rate per cell for each mouse
mean_eventrate_rows = []

for mouse, df in events_per_second_with_velocity_by_mouse.items():
    print(mouse)
    print('binning')
    
    cell_cols = [c for c in df.columns if str(c).startswith("cell_")]
    mean_events_per_cell = df[cell_cols].mean(axis=0, skipna=True)

    for cell_name, mean_val in mean_events_per_cell.items():
        mean_eventrate_rows.append({
            "mouse": mouse,
            "cell": cell_name,
            "mean_events_per_1s_window": mean_val,
        })

mean_eventrate_per_cell_df = pd.DataFrame(mean_eventrate_rows)
mean_eventrate_per_cell_df.head()
```
